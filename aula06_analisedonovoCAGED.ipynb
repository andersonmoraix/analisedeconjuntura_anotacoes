```python
# Instalar bibliotecas
!pip install ipeadatapy

# Importar bibliotecas
import ipeadatapy as ipea
import pandas as pd
```

# Dados

```python
# Obter metadados
metadados = ipea.metadata() #para saber todos códigos do IPEADATA
metadados
```

```python
# Identificar o código do CAGED
metadados[metadados.NAME.str.contains("Caged")] #procurar dados do CAGED, colocar o "C" maiusculo funcao sensivel a letras minusculas e maiusculas
```

```python
# Coletar dados
dados_admissoes = ipea.timeseries("CAGED12_ADMISN12") #admiditidos
dados_demissoes = ipea.timeseries("CAGED12_DESLIGN12") #demitidos
dados_saldo = ipea.timeseries("CAGED12_SALDON12") #saldo

dados_saldo.tail()
```

```python
# Tratamento de dados
(
    pd.concat([dados_admissoes, dados_demissoes, dados_saldo]) #para empilhar uma tabela em cima da outra
    .pivot(columns = "CODE", values = "VALUE (Pessoa)")
    .reset_index()
    .rename(
        columns = {
            "DATE": "data",
            "CAGED12_ADMISN12": "admissoes",
            "CAGED12_DESLIGN12": "demissoes",
            "CAGED12_SALDON12": "saldo"
            }
        )
).tail()
```
